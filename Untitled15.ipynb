```python
# ================= INSTALL DEPENDENCIES =================
!pip -q install sentence-transformers pytesseract pdf2image python-docx opencv-python PyPDF2
!apt-get -qq install -y tesseract-ocr
!apt-get -qq install -y poppler-utils

# ================= IMPORTS =================
import os
import numpy as np
import pytesseract
import cv2
import docx
from PyPDF2 import PdfReader
from pdf2image import convert_from_path
from sentence_transformers import SentenceTransformer
from google.colab import files

# ================= LOAD MODEL =================
print("🔄 Loading AI Model...")

try:
    model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
    print("✅ Model Loaded Successfully\n")
except Exception as e:
    print("❌ Error Loading Model:", e)
    raise

# ================= TEXT EXTRACTION =================

def extract_text_from_image(image_path):
    try:
        img = cv2.imread(image_path)

        if img is None:
            return ""

        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

        text = pytesseract.image_to_string(gray)

        return text.strip()

    except Exception as e:
        print(f"❌ Image OCR Error ({image_path}): {e}")
        return ""


def extract_text_from_pdf(path):
    text = ""

    try:
        # Try normal PDF extraction first
        reader = PdfReader(path)

        for page in reader.pages:
            extracted = page.extract_text()

            if extracted:
                text += extracted + "\n"

        # If text found, return
        if text.strip():
            return text.strip()

    except Exception as e:
        print(f"⚠ PDF text extraction failed: {e}")

    # OCR fallback
    try:
        pages = convert_from_path(path)

        for i, page in enumerate(pages):
            img_path = f"temp_page_{i}.jpg"

            page.save(img_path, "JPEG")

            text += extract_text_from_image(img_path) + "\n"

            if os.path.exists(img_path):
                os.remove(img_path)

        return text.strip()

    except Exception as e:
        print(f"❌ PDF OCR Error ({path}): {e}")
        return ""


def extract_text_from_docx(path):
    try:
        doc = docx.Document(path)

        text = "\n".join([para.text for para in doc.paragraphs])

        return text.strip()

    except Exception as e:
        print(f"❌ DOCX Error ({path}): {e}")
        return ""


def extract_text(path):
    ext = path.lower().split(".")[-1]

    if ext == "pdf":
        return extract_text_from_pdf(path)

    elif ext == "docx":
        return extract_text_from_docx(path)

    elif ext in ["jpg", "jpeg", "png"]:
        return extract_text_from_image(path)

    else:
        print(f"⚠ Unsupported File Type: {path}")
        return ""


# ================= CREATE EMBEDDING =================

def create_embedding(text):

    try:
        if not isinstance(text, str):
            text = str(text)

        text = text.strip()

        # Empty file handling
        if len(text) == 0:
            return np.zeros(384)

        # Limit text size
        text = text[:5000]

        embedding = model.encode(text)

        return embedding

    except Exception as e:
        print(f"❌ Embedding Error: {e}")
        return np.zeros(384)


# ================= COSINE SIMILARITY =================

def cosine_similarity(a, b):

    try:
        if np.linalg.norm(a) == 0 or np.linalg.norm(b) == 0:
            return 0.0

        similarity = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

        return float(similarity)

    except Exception as e:
        print(f"❌ Similarity Error: {e}")
        return 0.0


# ================= UPLOAD FILES =================

print("📂 Upload at least TWO files (PDF / DOCX / Images)\n")

uploaded = files.upload()

file_paths = list(uploaded.keys())

# ================= VALIDATION =================

if len(file_paths) < 2:
    print("❌ Please upload at least TWO files")

else:

    texts = {}
    embeddings = {}

    print("\n🔄 Extracting text...\n")

    for f in file_paths:

        text = extract_text(f)

        if len(text.strip()) == 0:
            print(f"⚠ No text found in: {f}")

        texts[f] = text

        embeddings[f] = create_embedding(text)

        print(f"✅ Processed: {f}")

    print("\n🔍 Comparing files...\n")

    threshold = 90  # percentage

    for i in range(len(file_paths)):

        for j in range(i + 1, len(file_paths)):

            f1 = file_paths[i]
            f2 = file_paths[j]

            sim = cosine_similarity(
                embeddings[f1],
                embeddings[f2]
            )

            percent = round(sim * 100, 2)

            print(f"\n🆚 {f1}  VS  {f2}")
            print(f"📊 Similarity: {percent}%")

            if percent >= threshold:
                print("✅ DUPLICATE FILES")

            else:
                print("❌ NOT DUPLICATE")

    print("\n✔ DONE")
```
